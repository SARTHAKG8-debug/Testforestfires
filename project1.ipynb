```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Algerian_forest_fires_dataset_UPDATE.csv',header=1)
df.head()
```

```python
df.info()
```

```python
df.isnull().sum() ##column wise null values
```

```python
df[df.isnull().any(axis=1)] ##row wise null values
```

```python
df.loc[:122,'region']=0
df.loc[122:,'region']=1
df[['region']]=df[['region']].astype('int')
df.head()
```

```python
df=df.dropna().reset_index(drop=True)
df.isnull().sum()
```

```python
df=df.drop(122).reset_index(drop=True)
df.iloc[[122]]
```

```python

df[['day','month','year','region','Temperature',' RH',' Ws']]=df[['day','month','year','region','Temperature',' RH',' Ws']].astype('int')
df.info()
```

```python
df.columns
```

```python
df[['Rain ', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']]=df[['Rain ', 'FFMC','DMC', 'DC', 'ISI', 'BUI', 'FWI']].astype('float')
df.info()
```

```python
df_copy=df.drop(['day','month','year'],axis=1)
df_copy.head()
```

```python

df_copy[['Classes  ']]=df_copy[['Classes  ']].astype('string')
df_copy.info()
```

```python
#encoding categorical features
df_copy['Classes  ']=np.where(df_copy['Classes  '].str.contains('not fire'),1,0)
df_copy.head()
```
